```python
# Cell 2: Config - PASTE YOUR KEY HERE
ROBOFLOW_API_KEY = 'YOUR_API_KEY_HERE'

DATASETS = [
    {'workspace':'tirths-workspace-k51ep', 'project':'mobile-damage-segmentation', 'version':1, 'slug':'mobile-damage', 'trade':'Phone Repair'},
    {'workspace':'bayram-grbz', 'project':'pipe-damage-detection-9eudj', 'version':1, 'slug':'pipe-damage', 'trade':'Plumber'},
    {'workspace':'sairohith', 'project':'complaint-management', 'version':1, 'slug':'complaint-management', 'trade':'AC/Electrical'},
    {'workspace':'bhoomikas-workspace-bdi6e', 'project':'smart-estate-carpentry', 'version':1, 'slug':'smart-estate-carpentry', 'trade':'Carpentry'},
]
EPOCHS=40; IMG_SIZE=640; BASE_MODEL='yolov8n.pt'
print('Config OK')
```

```python
# Cell 3: Download datasets
import os
from roboflow import Roboflow

rf = Roboflow(api_key=ROBOFLOW_API_KEY)
os.makedirs('datasets', exist_ok=True)
dataset_paths = {}

for ds in DATASETS:
    print(f'Downloading {ds["project"]}...')
    for v in [ds['version'], ds['version']+1, 1, 2, 3]:
        try:
            project = rf.workspace(ds['workspace']).project(ds['project'])
            dataset = project.version(v).download('yolov8', location=f'datasets/{ds["slug"]}')
            dataset_paths[ds['slug']] = dataset.location
            print(f'  OK: v{v} -> {dataset.location}')
            break
        except Exception as e:
            print(f'  v{v} failed: {e}')
            continue

print(f'Downloaded {len(dataset_paths)}/{len(DATASETS)}')
```

```python
# Cell 4: Train all models
from ultralytics import YOLO
import json, shutil, os

os.makedirs('trained_models', exist_ok=True)
training_results = {}

for ds in DATASETS:
    slug = ds['slug']
    if slug not in dataset_paths:
        print(f'Skipping {slug}')
        continue

    data_yaml = ''
    for root, dirs, files in os.walk(dataset_paths[slug]):
        for f in files:
            if f == 'data.yaml':
                data_yaml = os.path.join(root, f)
                break

    print(f'\nTraining {slug} ({ds["trade"]}) | yaml={data_yaml}')
    model = YOLO(BASE_MODEL)
    results = model.train(
        data=data_yaml, epochs=EPOCHS, imgsz=IMG_SIZE,
        project=f'runs/{slug}', name='train',
        device=0 if torch.cuda.is_available() else 'cpu',
        batch=16, patience=10, verbose=False, save=True
    )

    best_pt = f'runs/{slug}/train/weights/best.pt'
    output_pt = f'trained_models/{slug}.pt'
    if os.path.exists(best_pt):
        shutil.copy2(best_pt, output_pt)
        m2 = YOLO(output_pt)
        training_results[slug] = {
            'trade': ds['trade'], 'model_file': f'{slug}.pt',
            'classes': list(m2.names.values()),
            'map50': float(results.results_dict.get('metrics/mAP50(B)', 0))
        }
        print(f'  Saved {output_pt} | classes={training_results[slug]["classes"]} | mAP50={training_results[slug]["map50"]:.3f}')

with open('trained_models/models_metadata.json', 'w') as f:
    json.dump(training_results, f, indent=2)
print(json.dumps(training_results, indent=2))
```

```python
# Cell 5: Validate models
from ultralytics import YOLO
import os
for slug in training_results:
    pt = f'trained_models/{slug}.pt'
    if os.path.exists(pt):
        m = YOLO(pt)
        print(f'  {slug}.pt | {os.path.getsize(pt)/1e6:.1f}MB | {list(m.names.values())}')
```

```python
# Cell 6: Download ZIP
import zipfile, os
from google.colab import files

ZIP = 'fixitnow_yolov8_models.zip'
with zipfile.ZipFile(ZIP, 'w', zipfile.ZIP_DEFLATED) as zf:
    for f in os.listdir('trained_models'):
        p = os.path.join('trained_models', f)
        if os.path.isfile(p):
            zf.write(p, f)
            print(f'  Added {f} ({os.path.getsize(p)/1e6:.1f}MB)')
print(f'ZIP: {os.path.getsize(ZIP)/1e6:.1f}MB')
files.download(ZIP)
```
